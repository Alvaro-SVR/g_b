```python
import numpy as np
import scqubits as scq
import matplotlib.pyplot as plt
import imageio

import TransmonMap as tr_map
import TunableTransmonGIF as trtun_map
```

## Transmon

```python
tm = tr_map.TransmonMap(
    EC=1.0,
    ncut=30,
    ng_values=np.linspace(0.0, 1.0, 300),
    ratio_values=np.linspace(0,15,300)
)
tm.compute_maps()
tm.plot_map("N_map")
tm.plot_map("cos_map")
tm.plot_cut_ratio(12)
```

## Tunable Transmon

```python
tm = trtun_map.TunableTransmonGIF(
    EC=1.0,
    ncut=30,
    ng_values=np.linspace(0, 1.0, 120),
    ratio_values=np.linspace(0, 20, 160),
)

phi_vals = np.linspace(0, 1.0, 100)

path_with, path_without = tm.generate_gifs(
    phi_values=phi_vals,
    key="N_map",
    duration_seconds=10
)
```
