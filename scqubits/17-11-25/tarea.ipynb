```python
import numpy as np
import scqubits as scq
import matplotlib.pyplot as plt
import imageio

import TransmonMap as tr_map
import TunableTransmonGIF as trtun_map
import FluxoniumGIF as flux_map
```

## Transmon

```python
tm = tr_map.TransmonMap(
    EC=1.0,
    ncut=30,
    ng_values=np.linspace(0.0, 1.0, 300),
    ratio_values=np.linspace(0,15,300)
)

tm.compute_maps()

tm.plot_map(key="N_map", show_extrema=True, save=True)
tm.plot_map(key="cos_map", show_extrema=True, save=True)
tm.plot_cut_ratio(R_index=12, key="N_map", save=True)
```

## Tunable Transmon

```python
tm = trtun_map.TunableTransmonGIF(
    EC=1.0,
    ncut=30,
    ng_values=np.linspace(0, 1.0, 100),
    ratio_values=np.linspace(0, 20, 100),
    phi_values=np.linspace(0, 1.0, 100)
)

tm.generate_gifs_N(duration_seconds=10)
tm.generate_gifs_cos_phi(duration_seconds=10)
```
